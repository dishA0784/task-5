```python
import pandas as pd

sales = pd.read_csv("sales_data.csv")
customers = pd.read_csv("customer_churn.csv")

print(sales.head())
print(customers.info())
print(sales.isnull().sum())
```

```python
# Convert date column
sales['Date'] = pd.to_datetime(sales['Date'])

# Handle missing values
sales.fillna(0, inplace=True)

# Create new column
sales['Total_Sales'] = sales['Quantity'] * sales['Price']
```

```python
#df = pd.merge(sales, customers, on="Customer_ID", how="inner")
df = pd.merge(sales, customers,
              left_on='Customer_ID',
              right_on='CustomerID',
              how='inner')
print(df.head())
```

```python
sales['Total_Sales'] = sales['Quantity'] * sales['Price']
```

```python
try:
    df = pd.merge(sales, customers, on="Customer_ID", how="inner")
    
    # If merge is empty → fallback
    if df.shape[0] == 0:
        df = sales.copy()
except:
    df = sales.copy()
```

```python
# Total Revenue
total_revenue = df['Total_Sales'].sum()

# Total Customers
if 'Customer_ID' in df.columns:
    total_customers = df['Customer_ID'].nunique()
else:
    total_customers = 0

# Average Order Value
avg_order_value = df['Total_Sales'].mean()

# Top Customer
if 'Customer_ID' in df.columns:
    top_customer_data = df.groupby('Customer_ID')['Total_Sales'].sum().sort_values(ascending=False)
    top_customer_id = top_customer_data.index[0]
    top_customer_value = top_customer_data.iloc[0]
else:
    top_customer_id = "N/A"
    top_customer_value = 0
```

```python
print("CUSTOMER SALES ANALYSIS REPORT\n")

print(f"Total Revenue: ${total_revenue:,.0f}")
print(f"Total Customers: {total_customers:,}")
print(f"Average Order Value: ${avg_order_value:,.0f}")
print(f"Top Customer: Customer_ID {top_customer_id} - ${top_customer_value:,.0f}")
```
